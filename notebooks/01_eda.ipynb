```python
import pandas as pd
```

```python
#Path to LZ (load zone) and HB (hub) Settlement price point (SPP) 2024 data
dam2024_path = "../data/raw/DAMLZHBSPP_2024.xlsx"
```

**Exploring the data and understanding the columns**

March and november data includes changes in daylight savings time. The finding are noted in notes/story.md in the project files.

```python
#Exploring March and November data for Daylight savings time adjustments
mar = pd.read_excel(dam2024_path, sheet_name="Mar")
#mar["Settlement Point"].value_counts()

nov = pd.read_excel(dam2024_path, sheet_name="Nov")
# nov["Settlement Point"].value_counts()

# nov[nov["Repeated Hour Flag"] == "Y"]
```

*Combining all sheets into a single dataframe*

```python
sheets = pd.read_excel(dam2024_path, sheet_name=None)
dam2024_df = pd.concat(sheets.values(), ignore_index=True)

#check- should be 366 (2024 is a leap year)
dam2024_df["Delivery Date"].nunique()
```

*Average SPP for the year*

```python
avg_price = dam2024_df["Settlement Point Price"].mean().item()
print(f"Average price: ${avg_price:,.2f}/MWh")
```

*Maximum Price of the year (Same can be done with minimum)*

```python
max_price = dam2024_df["Settlement Point Price"].max().item()
print("Maximum Settlement Point Price: $", max_price, "/ MWh")

max_price_location = dam2024_df.loc[dam2024_df["Settlement Point Price"].idxmax()]
print("Location of Maximum Settlement Point Price:")
print(max_price_location)
```

*When did the price go negative?*

```python
neg = dam2024_df[dam2024_df["Settlement Point Price"] < 0]
neg[["Delivery Date", "Hour Ending", "Settlement Point", "Settlement Point Price"]].sort_values(by="Settlement Point Price", ascending=True).head(20)
```

*Which hub is the cheapest?*

```python
#cheapest average hub
cheapest_hub = dam2024_df.groupby("Settlement Point")["Settlement Point Price"].mean()
print (cheapest_hub.sort_values(ascending=True))
cheapest_hub_location = cheapest_hub.idxmin()
print("Cheapest Hub Location:", cheapest_hub_location)
```
